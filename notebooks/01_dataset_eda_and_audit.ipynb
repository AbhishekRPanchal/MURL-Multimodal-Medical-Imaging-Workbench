# 01. Breast Ultrasound Dataset (BUSI) Audit & Exploratory Data Analysis
### MURL — Multimodal Medical Imaging Research Workbench

This notebook demonstrates rigorous dataset provenance analysis, image-mask pairing verification, multi-mask fusion, and leak-free stratified split generation.

```python
import os
import json
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load audited metadata
with open('data/processed/dataset_metadata.json') as f:
    meta = json.load(f)

print('Provenance:')
for k, v in meta['provenance'].items():
    print(f'  {k}: {v}')

print('\nAudit Summary:')
for k, v in meta['audit_summary'].items():
    print(f'  {k}: {v}')
```

```python
# Visualize class distributions across stratified splits
classes = ['benign', 'malignant', 'normal']
train_counts = [sum(1 for c in meta['cases'] if c['split'] == 'train' and c['class'] == cls) for cls in classes]
val_counts = [sum(1 for c in meta['cases'] if c['split'] == 'val' and c['class'] == cls) for cls in classes]
test_counts = [sum(1 for c in meta['cases'] if c['split'] == 'test' and c['class'] == cls) for cls in classes]

x = np.arange(len(classes))
width = 0.25

plt.figure(figsize=(8, 5))
plt.bar(x - width, train_counts, width, label='Train (70%)', color='#2563eb')
plt.bar(x, val_counts, width, label='Val (15%)', color='#f59e0b')
plt.bar(x + width, test_counts, width, label='Held-out Test (15%)', color='#10b981')

plt.xlabel('Diagnostic Category')
plt.ylabel('Number of Cases')
plt.title('Case-Level Stratified Partitioning (BUSI)')
plt.xticks(x, [c.capitalize() for c in classes])
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```
