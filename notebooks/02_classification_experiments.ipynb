# 02. Ultrasound Classification Experiments & Benchmark Comparison
### MURL — Multimodal Medical Imaging Research Workbench

Empirical comparison between custom UltrasoundBaselineCNN and Pretrained ResNet18 Transfer Learning with Class-Weighted Objectives.

```python
import os
import json
import matplotlib.pyplot as plt

# Load test evaluation results
with open('outputs/test_evaluation_report.json') as f:
    report = json.load(f)

cls_rep = report['classification']
for model_name, metrics in cls_rep.items():
    print(f'=== {model_name.upper()} ===')
    print(f'  Test Accuracy:    {metrics["accuracy"]*100:.2f}%')
    print(f'  Macro Precision:  {metrics["precision_macro"]:.4f}')
    print(f'  Macro Recall:     {metrics["recall_macro"]:.4f}')
    print(f'  Macro F1-Score:   {metrics["f1_macro"]:.4f}')
    print(f'  ROC-AUC (Macro):  {metrics["roc_auc_macro"]:.4f}')
```
